```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("winequality_red.csv", sep=';')
df.head()
```

#Load the dataset and handle any missing data by replacing null values with the mean value of the respective column. (Score 2)

```python
(df == 0).sum()
```

```python
c_acid_mean = df['citric acid'].mean().round(2)
df['citric acid']=df['citric acid'].replace(0, c_acid_mean)
print((df == 0).sum())
print(df.head())
```

Extract the following columns as vectors: alcohol,citric acid. (Score 2)

```python
alc_vector = df['alcohol']
c_acid_vector = df['citric acid']
print("Alcohol vectors\n",alc_vector)
print("Citric acid vectors\n",c_acid_vector)
```

Select two features (e.g., alcohol and density) from the dataset and calculate the
covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the
selected columns.(Score 2)

```python
x= df['alcohol']
t = df['density']
cov_matrix = np.cov(x.T,t)
print(cov_matrix)
print("Covariance < 0 → one increases while the other decreases.")
```

Perform eigen decomposition on the covariance matrix you computed in question 3.
Identify and interpret the results:Identify the top 2 eigenvalues of the covariance
matrix,Identify the corresponding eigenvectors.(Score 2)

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print("Eigenvalues :", eigenvalues)
print("Eigenvectors :",eigenvectors)
```
