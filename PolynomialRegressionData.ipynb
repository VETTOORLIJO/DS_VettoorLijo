```python
X=df.drop('Ice Cream Sales (units)',axis=1)
y=df['Ice Cream Sales (units)']
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
pr=PolynomialFeatures(degree=2)
X_poly = pr.fit_transform(X)
```

```python
lr=LinearRegression()
lr.fit(X_poly,y)
```

```python
y_prd = lr.predict(X_poly)
```

```python
lr.coef_
```

```python
plt.scatter(X,y)
plt.plot(X,y_prd,color='red')
plt.show()
#
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
mse=mean_squared_error(y,y_prd)
mae=mean_absolute_error(y,y_prd)
r2=r2_score(y,y_prd)
print('MSE:',mse)
print('MAE:',mae)
print('R2:',r2)
```

```python
pr1=PolynomialFeatures(degree=3)
X_poly = pr1.fit_transform(X)
lr=LinearRegression()
lr.fit(X_poly,y)
```

```python
y_prd = lr.predict(X_poly)
```

```python
plt.scatter(X,y)
plt.plot(X,y_prd,color='red')
plt.show()
```
