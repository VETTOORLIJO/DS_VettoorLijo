```python
df= pd.read_excel("ABC_Company.xlsx")
df
```

```python
#"height" column by replacing it with random numbers between 150 and 180.
df["Height"] =np.random.randint(80, 100, size=len(df))
print(df["Height"])
```

```python
# employe percentage split
df["Name"].duplicated().sum()
emp_team_split=((df["Team"].value_counts()/len(df["Name"]))
                *100).round(2)
print(emp_team_split)
plt.pie(
     emp_team_split,
     autopct='%1.1f%%',
     pctdistance=1.15,     # pushes % outside
     labeldistance=1.05    # pushes labels slightly outside
 )

plt.show()
```

```python
# employs distribution of employees across each team is almost same expet two teams
```

```python
#Segregate positions within the company
print(df["Position"].isna().sum())
seg_position = df.sort_values(["Position", "Name"]).reset_index()
print(seg_position[["Name", "Position"]])
position_count =  seg_position[ "Position"].value_counts()
sns.barplot(position_count)
plt.show()
```

```python
#The majority of employees are concentrated in the SG position, indicating it is the most common role in the dataset.
# In contrast, the C position has the lowest employee count, making it the least represented role.
```

```python
# predominant age group
emp_age = df["Age"].value_counts()
print(emp_age)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.kdeplot(df["Age"])
plt.title("Age KDE")
plt.grid(True, linestyle="--", alpha=0.6)
plt.subplot(1,2,2)
sns.lineplot(x=emp_age.index, y=emp_age.values)
plt.title("Age Frequency")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
# Most employees fall within the 23–28 age range, forming a distribution that closely resembles a bell curve, indicating an approximately normal age distribution in the dataset.
```

```python
print(df["Salary"].isna().sum())
print(df[df["Salary"]<=500000])
```

```python
#highest salary position and team
print(df.Team[df["Salary"].isna()].reset_index(drop=True))
df_grouped=df.groupby("Salary")[["Team","Position"]].max().reset_index()
print(df_grouped)
sns.scatterplot(
    x="Salary",
    y="Team",
    hue="Position",
    data=df_grouped,
    palette="tab10",
)
plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))
plt.grid(True, linestyle="--", alpha=0.6)   # add grid
plt.show()

```

```python
#Across the organisation, a large proportion of employees earn salaries in the 100,000 to 500,000 range,
#Within the Los Angeles Lakers, employees in the SF position receive comparatively higher pay.
```

```python
df_age_sal = df[["Age", "Salary"]].sort_values("Salary", ascending=False)
print(df_age_sal)
sns.lineplot(x="Age", y="Salary",data=df_age_sal)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
df_corrr = df[["Age", "Salary"]].corr()
print(df_corrr)
sns.heatmap(df_corrr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# The plot and heatmap together show that the relationship between Age and Salary is non‑monotonic
# (it doesn’t consistently increase or decrease). Salaries fluctuate across ages, with peaks and dips, rather than following a smooth upward or downward trend.
#The correlation coefficient of ≈ 0.21 confirms that the relationship is weakly positive —
#meaning older employees tend to earn slightly more, but age isn’t a strong predictor of salary in this dataset.
```
