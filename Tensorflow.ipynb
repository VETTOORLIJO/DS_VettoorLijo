```python
#changing the type of data
a=tf.constant(3.12345678,tf.float32)
b=tf.cast(a,dtype=tf.int32)
print(b)
print(a.dtype)
print(b.dtype)
```

```python
x=tf.constant([1,2,3])
y=tf.constant([5,6,7])
```
